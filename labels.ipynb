# Unification of labels
In this file we try to include in a single folder and a single file for each image both the tool labels and the tti labels.

At the end, inside the yolo_dataset you'll have a new folder called labels_combined. It is important, in order to train the Yolo model, to modify the name of the folder in labels and ensure to have a single folder named so.

### Imports

```python
from pathlib import Path
```

### Dataset root

```python
YOLO_ROOT = "yolo_dataset"
```

### Mix!

```python
def merge_yolo_labels(dataset_root):
    dataset_root = Path(dataset_root)
    
    for split in ["train", "val", "test"]:
        labels_dir = dataset_root / "labels" / split
        tti_dir = dataset_root / "tti_labels" / split
        out_dir = dataset_root / "labels_combined" / split
        out_dir.mkdir(parents=True, exist_ok=True)
        
        # Raccogli tutti i nomi di file unici presenti in entrambe le cartelle
        all_txt_files = set([f.name for f in labels_dir.glob("*.txt")]) | \
                        set([f.name for f in tti_dir.glob("*.txt")])
        
        for txt_name in all_txt_files:
            combined_content = []
            
            # Leggi annotazioni tool
            tool_file = labels_dir / txt_name
            if tool_file.exists():
                with open(tool_file, "r") as f:
                    combined_content.extend(f.readlines())
                    
            # Leggi annotazioni TTI
            tti_file = tti_dir / txt_name
            if tti_file.exists():
                with open(tti_file, "r") as f:
                    combined_content.extend(f.readlines())
            
            # Salva il file combinato
            with open(out_dir / txt_name, "w") as f:
                f.writelines(combined_content)
```

```python
# Esegui la fusione delle etichette
merge_yolo_labels(YOLO_ROOT)
print("Etichette fuse con successo in 'labels_combined'!")
```
